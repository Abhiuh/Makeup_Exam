```python
import pandas as pd
import numpy as np

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.utils import to_categorical


df = pd.read_csv("next_word_prediction_dataset.csv")


texts = df['text'].str.lower()


tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

word_index = tokenizer.word_index
vocab_size = len(word_index) + 1

print("Vocabulary Size:", vocab_size)


input_sequences = []

for line in texts:

    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i + 1]
        input_sequences.append(n_gram_sequence)


max_seq_len = max([len(seq) for seq in input_sequences])

input_sequences = np.array(
    pad_sequences(input_sequences,
                  maxlen=max_seq_len,
                  padding='pre')
)

print("Maximum Sequence Length:", max_seq_len)


X = input_sequences[:, :-1]
y = input_sequences[:, -1]


y = to_categorical(y, num_classes=vocab_size)

print("Input Shape:", X.shape)
print("Output Shape:", y.shape)

model = Sequential()


model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim=100,
        input_length=max_seq_len - 1
    )
)


model.add(LSTM(150))

model.add(Dense(vocab_size, activation='softmax'))


model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)


model.summary()


history = model.fit(
    X,
    y,
    epochs=50,
    batch_size=32,
    verbose=1
)


model.save("next_word_lstm_model.h5")

print("\nModel Training Completed Successfully!")
print("Model Saved as: next_word_lstm_model.h5")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import load_model


df = pd.read_csv("next_word_prediction_dataset.csv")

texts = df['text'].str.lower()


tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

word_index = tokenizer.word_index
vocab_size = len(word_index) + 1


input_sequences = []

for line in texts:

    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i + 1]
        input_sequences.append(n_gram_sequence)


max_seq_len = max([len(seq) for seq in input_sequences])


model = load_model("next_word_lstm_model.h5")


plt.figure(figsize=(8,5))

plt.plot(history.history['accuracy'])

plt.title("Training Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")

plt.show()

#
plt.figure(figsize=(8,5))

plt.plot(history.history['loss'])

plt.title("Training Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")

plt.show()


def predict_top3_words(seed_text):

    token_list = tokenizer.texts_to_sequences([seed_text.lower()])[0]

    token_list = pad_sequences(
        [token_list],
        maxlen=max_seq_len - 1,
        padding='pre'
    )

    prediction = model.predict(token_list, verbose=0)[0]

    # Get top 3 indices
    top_indices = prediction.argsort()[-3:][::-1]

    predicted_words = []

    for index in top_indices:

        for word, idx in tokenizer.word_index.items():

            if idx == index:
                predicted_words.append(word)

    return predicted_words


def autocomplete_sentence(seed_text, num_words=5):

    for _ in range(num_words):

        token_list = tokenizer.texts_to_sequences([seed_text.lower()])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_seq_len - 1,
            padding='pre'
        )

        prediction = model.predict(token_list, verbose=0)

        predicted_index = np.argmax(prediction)

        output_word = ""

        for word, index in tokenizer.word_index.items():

            if index == predicted_index:
                output_word = word
                break

        seed_text += " " + output_word

    return seed_text


seed_text = input("Enter Sentence Fragment: ")

# Top-3 Predictions
top3 = predict_top3_words(seed_text)

print("\nTop-3 Predicted Words:")
for i, word in enumerate(top3, 1):
    print(f"{i}. {word}")

# Auto Completion
completed_sentence = autocomplete_sentence(seed_text, num_words=5)

print("\nAuto Completed Sentence:")
print(completed_sentence)
```
